```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, _ = load_iris(return_X_y=True)

modelo_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

grupos_previstos = modelo_kmeans.fit_predict(X)

silhueta = silhouette_score(X, grupos_previstos)
print(f"Coeficiente de Silhouette (Qualidade da separação): {silhueta:.3f}")

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=grupos_previstos, cmap='viridis', edgecolors='k', s=60)

centroides = modelo_kmeans.cluster_centers_
plt.scatter(centroides[:, 0], centroides[:, 1], c='red', marker='X', s=200, label='Centróides')
plt.title('Agrupamento K-Means - Dados Iris')
plt.legend()
plt.show()
```
