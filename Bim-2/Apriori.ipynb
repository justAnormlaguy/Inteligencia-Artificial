```python
import pandas as pd
from sklearn.datasets import load_iris
from mlxtend.frequent_patterns import apriori, association_rules

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

df_binario = df > df.mean()

itemsets_frequentes = apriori(df_binario, min_support=0.3, use_colnames=True)

regras = association_rules(itemsets_frequentes, metric="confidence", min_threshold=0.7)

print("Exemplos de Regras de Associação Encontradas (Características Florais):")
display(regras[['antecedents', 'consequents', 'support', 'confidence']].head())
```
